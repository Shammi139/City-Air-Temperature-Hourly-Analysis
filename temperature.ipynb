```python
# Libraries used
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load data
df = pd.read_csv("temperature_analysis.csv")
```

```python
# Statistics
print("Mean Temperature:", df["Temperature"].mean())
print("Variance:", df["Temperature"].var())
```

```python
# Line plot
plt.plot(df["Hour"], df["Temperature"])
plt.xlabel("Hour")
plt.ylabel("Temperature (°C)")
plt.title("City Air Temperature Hourly Analysis")
plt.show()
```
